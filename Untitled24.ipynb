```python
# Importando as bibliotecas
import pandas as pd
import sqlite3
import datetime as dt

# Carregando os dados limpos do banco de dados
conn = sqlite3.connect('online_retail.db')
df = pd.read_sql_query("SELECT * FROM online_retail;", conn)
conn.close()

# 1. Limpeza de dados
df.dropna(subset=['CustomerID'], inplace=True)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df = df[df['Quantity'] > 0]

# 2. Engenharia de Atributos - Criando a coluna 'Monetary'
df['Monetary'] = df['Quantity'] * df['UnitPrice']

# 3. Calculando a Recência
ultima_data = df['InvoiceDate'].max()
data_referencia = ultima_data + dt.timedelta(days=1)

# 4. Agrupando os dados para calcular o RFM
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda InvoiceDate: (data_referencia - InvoiceDate.max()).days,
    'InvoiceNo': lambda InvoiceNo: len(InvoiceNo.unique()),
    'Monetary': lambda Monetary: Monetary.sum()
})

# 5. Renomeando as colunas
rfm.columns = ['Recency', 'Frequency', 'Monetary']

# Exibindo as primeiras 5 linhas da nova tabela RFM
print("Tabela RFM (Recência, Frequência, Monetariedade) dos clientes:")
print(rfm.head())
```

```python
# Importando as bibliotecas necessárias
import pandas as pd
import sqlite3
import datetime as dt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 1. Carregar e preparar os dados RFM (código da etapa anterior)
conn = sqlite3.connect('online_retail.db')
df = pd.read_sql_query("SELECT * FROM online_retail;", conn)
conn.close()
df.dropna(subset=['CustomerID'], inplace=True)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df = df[df['Quantity'] > 0]
df['Monetary'] = df['Quantity'] * df['UnitPrice']
ultima_data = df['InvoiceDate'].max()
data_referencia = ultima_data + dt.timedelta(days=1)
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda InvoiceDate: (data_referencia - InvoiceDate.max()).days,
    'InvoiceNo': lambda InvoiceNo: len(InvoiceNo.unique()),
    'Monetary': lambda Monetary: Monetary.sum()
})
rfm.columns = ['Recency', 'Frequency', 'Monetary']

# 2. Normalizando os dados RFM
rfm_scaled = rfm.copy()
scaler = StandardScaler()
rfm_scaled[['Recency', 'Frequency', 'Monetary']] = scaler.fit_transform(rfm_scaled[['Recency', 'Frequency', 'Monetary']])
print("Primeiras 5 linhas da tabela RFM normalizada:")
print(rfm_scaled.head())

# 3. Método do Cotovelo para encontrar o número de clusters
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(rfm_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Método do Cotovelo')
plt.xlabel('Número de Clusters (K)')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

```python
# Importando as bibliotecas necessárias
import pandas as pd
import sqlite3
import datetime as dt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Preparando os dados (mesmo código das etapas anteriores)
conn = sqlite3.connect('online_retail.db')
df = pd.read_sql_query("SELECT * FROM online_retail;", conn)
conn.close()

df.dropna(subset=['CustomerID'], inplace=True)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df = df[df['Quantity'] > 0]
df['Monetary'] = df['Quantity'] * df['UnitPrice']
ultima_data = df['InvoiceDate'].max()
data_referencia = ultima_data + dt.timedelta(days=1)

rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda InvoiceDate: (data_referencia - InvoiceDate.max()).days,
    'InvoiceNo': lambda InvoiceNo: len(InvoiceNo.unique()),
    'Monetary': lambda Monetary: Monetary.sum()
})
rfm.columns = ['Recency', 'Frequency', 'Monetary']

# 2. Normalizando os dados RFM
rfm_scaled = rfm.copy()
scaler = StandardScaler()
rfm_scaled[['Recency', 'Frequency', 'Monetary']] = scaler.fit_transform(rfm_scaled[['Recency', 'Frequency', 'Monetary']])

# 3. Aplicando o K-Means com 4 clusters
kmeans = KMeans(n_clusters=4, init='k-means++', random_state=42, n_init=10)
kmeans.fit(rfm_scaled)
rfm['Cluster'] = kmeans.labels_

# 4. Analisando a média de cada cluster para entender os segmentos
analise_clusters = rfm.groupby('Cluster').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': ['mean', 'count']
}).round(2)

print("Análise dos clusters (média de Recência, Frequência e Monetariedade):")
print(analise_clusters)

# 5. Exibindo a tabela RFM com os clusters para os primeiros 10 clientes
print("\nPrimeiros 10 clientes com o rótulo de cluster:")
print(rfm.head(10))
```
